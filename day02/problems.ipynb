```python
from helper import load_bike_raw
df = load_bike_raw()
```

```python
# date, hour, rented 뽑아서 출력 -> 상단 3개만
df[['date', 'hour', 'rented']].head(3)
df['functioning'] # No ->운영안함 | Yes 운영함 <- 각각의 sum, mean
off = df[df['functioning'] == 'No']
on = df[df['functioning'] == 'Yes']
print('[sum] off:', off.rented.sum(), 'on:', on.rented.sum())
```

```python
# 봄,가을 시간대만 AND 온도가 20~25도 <- 해당 하는 데이터를 뽑을것
# isin([..., ...]), .between(값A, 값B)
result = df[df['season'].isin(['Spring', 'Autumn']) & df['temp'].between(20, 25)]
display(result)

# 대여수가 가장 많았던 상위 5개 시간대를 뽑아보세요
# .nlargest
df.nlargest(5, 'rented')
```
