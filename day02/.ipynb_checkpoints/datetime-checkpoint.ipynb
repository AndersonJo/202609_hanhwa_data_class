```python
import pandas as pd
from helper import load_bike_raw
```

```python
df = load_bike_raw()
print(df['date'].dtype)
df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d')
print(df['date'].dtype)
```

```python
# df['date'].dt.day_name() # .month, .dayofweek, .day_name()
```

```python
df['datetime'] = df['date'] + pd.to_timedelta(df['hour'], unit='h')
df[['date', 'hour', 'datetime', 'rented']].head(3)
```

```python
# df.groupby(df['date'].dt.month)['rented'].sum().plot()
# df['hour'] # <- 시간으로 groupby 하고 평균 rented 갯수 구하고 -> 오름차순 정렬 sort_values 사용 | .round(0)
# df.groupby('hour')['rented'].mean().round(0).sort_values(ascending=True)
```

```python
df['is_weekend'] = df['date'].dt.dayofweek >= 5
# 주말이냐, 주말이 아니냐로 groupby -> 평균값 \
df.groupby('is_weekend')['rented'].mean().round()
```

```python
# holiday 로 groupby -> rented 평균값 
df['holiday']
df.groupby('holiday')['rented'].mean()
```
