```python
import csv

with open(titanic_path(), encoding='utf-8') as f:
    reader = csv.DictReader(f)
    rows = list(reader)
print(rows[0]['Name'], rows[0]['Age'], rows[0]['Ticket'])
```

```python
first_class = []
for row in rows:
    if row['Pclass'] == '1':
        first_class.append({'Name': row['Name'], 'Fare': row['Fare']})

out = output_path('first_class.csv')
with open(out, 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['Name', 'Fare'])
    writer.writeheader()
    writer.writerows(first_class)
out.read_text(encoding='utf-8').splitlines()[1]
```

```python
import threading 
lock = threading.Lock()
log_path = output_path('log.txt')
open(log_path, 'w').close()

def write_lines(writer_name, count):
    for i in range(count):
        with lock: # A 진입 -> B, C 멈춤
            with open(log_path, 'a', encoding='utf-8') as f:
                f.write(f'[{writer_name}] line {i}\n')
threads = []
for name in ['A', 'B', 'C']:
    t = threading.Thread(target=write_lines, args=(name, 20))
    threads.append(t)
    t.start()
for t in threads:
    t.join()  # 멈춤
with open(log_path, encoding='utf-8') as f:
    lines = f.readlines()
    # for l in lines:
        # print(l)
```

```python
import csv

with open(titanic_path(), encoding='utf-8') as f:
    reader = csv.DictReader(f)
    rows = list(reader)
print(rows[0]['Name'], rows[0]['Age'], rows[0]['Ticket'])
# Problem: Age == '' empty string 인 갯수를 찾아내서 카운팅
missing = 0
for row in rows: 
    if row['Age'] == '':
        missing += 1
print('missing:', missing)
```

```python
# female_names.txt <- 만들고, female 사람의 Name 만 들어가도록 코딩
out = output_path('female_names.txt')
with open(out, 'w', encoding='utf-8') as f:
    for row in rows:
        if row['Sex'] == 'female':
            f.write(row['Name'] + '\n')
print(out.read_text(encoding='utf-8').splitlines())
```
